```r
library(lme4)
library(tidyverse)
packageVersion("lme4")
```

```r
# set path 
base_path = "/home/EOCRC_atlas/"
```

```r
# load meta cluster assignments 
df = read.csv(paste0(base_path, '/results/EOCRC_inferCNV/CNV_cluster_df.csv'))
head(df)
```

```r
# scale age 
df$age_scaled = scale(df$age)
```

```r
# set up data frame 
df_clean <- df %>%
  mutate(
    cluster = fct_relevel(as.factor(cluster), "1"),
    side    = fct_relevel(as.factor(side), "Left"),
    treat   = as.factor(treat),
    stage   = as.factor(stage),
    sex     = as.factor(sex)
  )

# run regressions 
# cluster 2 (high) vs cluster 1 (low)
df_c2_vs_c1 <- df_clean %>% filter(cluster %in% c("1", "2"))
fit_high_vs_low <- glmer(cluster ~ age_scaled + side + treat + stage + sex + (1 | patient_id),data = df_c2_vs_c1,family = binomial)
summary(fit_high_vs_low)

# cluster 3 (medium) vs cluster 1 (low)
df_c3_vs_c1 <- df_clean %>% filter(cluster %in% c("1", "3"))
fit_med_vs_low <- glmer(cluster ~ age_scaled + side + treat + stage + sex + (1 | patient_id),data = df_c3_vs_c1,family = binomial)
summary(fit_med_vs_low)
```
